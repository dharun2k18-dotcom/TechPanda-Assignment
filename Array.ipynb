```python
transactions = [5000, -2000, 3000, -1000, 7000, -1500]
```

```python
deposits = 0
withdrawals = 0

for amount in transactions:
    if amount > 0:
        deposits = deposits + amount
    else:
        withdrawals = withdrawals + amount

print("Total deposits:", deposits)
print("Total withdrawals:", withdrawals)



net = sum(transactions)
print("Net transaction amount:", net)



highest_deposit = 0
for amount in transactions:
    if amount > 0 and amount > highest_deposit:
        highest_deposit = amount
print("Highest deposit:", highest_deposit)



highest_withdrawal = 0
for amount in transactions:
    if amount < 0 and amount < highest_withdrawal:
        highest_withdrawal = amount
print("Highest withdrawal:", highest_withdrawal)



print("Number of transactions:", len(transactions))
```

```python
products = ["Laptop", "Mouse", "Keyboard", "Monitor", "Printer"]
```

```python
stock = [8, 25, 6, 15, 4]
```

```python
print("Products with stock below 10:")
for i in range(len(stock)):
    if stock[i] < 10:
        print(products[i], "-", stock[i])


print("Highest stock:", max(stock))


print("Lowest stock:", min(stock))


product = input("Enter product name to add stock: ")
quantity = int(input("Enter quantity to add: "))
position = products.index(product)
stock[position] = stock[position] + quantity
print("Updated stock:", stock)


product = input("Enter product name sold: ")
quantity = int(input("Enter quantity sold: "))
position = products.index(product)
if quantity <= stock[position]:
    stock[position] = stock[position] - quantity
    print("Sale completed")
else:
    print("Insufficient stock")



print("Total stock:", sum(stock))
```
