```python
data2 = {"Name": ["Arun", "Priya", "Kumar", "Divya", "Arun", None],
        "Age": [25, 30, np.nan, 35, 25, 28],
        "Salary": [35000, 45000, 40000, np.nan, 35000, 50000],
        "City": ["Chennai", "Mumbai", "Chennai", None, "Chennai", "Delhi"]}

df2 = pd.DataFrame(data2)

print(df2)
```

```python
df2_clean = df2.dropna()

print(df2_clean)
```

```python
df2_unique = df2.drop_duplicates()

print(df2_unique)
```

```python
result = df.groupby("Department")["Salary"].mean()

department = result.idxmax()
average_salary = result.max()

print("Department:", department)
print("Average Salary:", average_salary)
```

```python
df["SalaryAfterBonus"] = df["Salary"] + df["Bonus"]

print(df)
```
