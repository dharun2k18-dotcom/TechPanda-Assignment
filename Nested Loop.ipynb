```python
seats = [["Available", "Reserved", "Available"],
         ["Available", "Available", "Reserved"],
         ["Reserved", "Available", "Available"]]

requested = 4
allocated = 0

for row in seats:
    for seat in row:

        if seat == "Reserved":
            continue

        if seat == "Available":
            allocated += 1
            print("Seat allocated")

        if allocated == requested:
            break
    
    if allocated == requested:
        break

print("Total seats allocated:", allocated)
```

```python
warehouse = [[("Laptop", 5), ("Mouse", 0), ("Keyboard", 10)],[("Monitor", 0), ("Printer", 4), ("Laptop Bag", 8)],[("Cable", 0), ("Charger", 6), ("Tablet", 3)]]

target = "Printer"
found = False

for row in warehouse:
    for product, stock in row:

        if stock == 0:
            continue

        print(product, "Stock:", stock)

        if product == target:
            print("Product located")
            found = True
            break

    if found:
        break
```

```python
seats = [["Available", "Booked", "Available"],
         ["Booked", "Available", "Booked"],
         ["Available", "Booked", "Available"]]

requested_row = 2
requested_column = 2

booked = False

for row in range(3):
    for column in range(3):

        if seats[row][column] == "Booked":
            continue

        if row == requested_row and column  == requested_column:
            seats[row][column] = "Booked"
            print("Requested seat booked")
            booked = True
            break

    if booked:
        break

if not booked:
    print("Requested seat is not available")
```
