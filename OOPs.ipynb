```python
class Employee:
    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.salary = salary

    def display(self):
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Salary:", self.salary)


class Developer(Employee):
    def __init__(self, name, employee_id, salary, programming_language):
        super().__init__(name, employee_id, salary)
        self.programming_language = programming_language

    def display(self):
        super().display()
        print("Programming Language:", self.programming_language)


class Manager(Employee):
    def __init__(self, name, employee_id, salary, team_size):
        super().__init__(name, employee_id, salary)
        self.team_size = team_size

    def display(self):
        super().display()
        print("Team Size:", self.team_size)


developer1 = Developer("Dharun", 101, 50000, "Python")
manager1 = Manager("Arun", 102, 70000, 10)

print("Developer Details")
developer1.display()

print("Manager Details")
manager1.display()
```

```python
class Vehicle:
    def __init__(self, brand, model, rent_per_day):
        self.brand = brand
        self.model = model
        self.rent_per_day = rent_per_day

    def calculate_rent(self, days):
        return self.rent_per_day * days


class Car(Vehicle):
    def calculate_rent(self, days):
        rent = self.rent_per_day * days
        insurance = rent * 10 / 100
        return rent + insurance


class Bike(Vehicle):
    def calculate_rent(self, days):
        rent = self.rent_per_day * days
        maintenance = rent * 5 / 100
        return rent + maintenance


car1 = Car("Toyota", "Innova", 2000)
bike1 = Bike("Yamaha", "R15", 1000)

print("Car Rent:", car1.calculate_rent(3))
print("Bike Rent:", bike1.calculate_rent(3))
```

```python
class BankAccount:
    def __init__(self, name, account_number, balance):
        self.name = name
        self.account_number = account_number
        self.__balance = balance                                                       #without __ balance can be accessed and modified directly from outside the class, 
                                                                                       #which can lead to unintended consequences. By using __balance, 
                                                                                       #we make it private and protect it from direct access.

    def deposit(self, amount):
        if amount > 0:
            self.__balance = self.__balance + amount
            print("Amount deposited successfully")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive")
        elif amount > self.__balance:
            print("Insufficient balance")
        else:
            self.__balance = self.__balance - amount
            print("Amount withdrawn successfully")

    def get_balance(self):
        return self.__balance


account1 = BankAccount("Dharun", 12345, 10000)

account1.deposit(5000)
account1.withdraw(3000)

print("Balance:", account1.get_balance())
```

```python
class UPIPayment:
    def make_payment(self, amount):
        print("Payment of", amount, "made using UPI")


class CreditCardPayment:
    def make_payment(self, amount):
        print("Payment of", amount, "made using Credit Card")


class NetBankingPayment:
    def make_payment(self, amount):
        print("Payment of", amount, "made using Net Banking")


def process_payment(payment_mode, amount):
    payment_mode.make_payment(amount)


upi = UPIPayment()
credit_card = CreditCardPayment()
net_banking = NetBankingPayment()

process_payment(upi, 500)
process_payment(credit_card, 1000)
process_payment(net_banking, 2000)
```

```python
from abc import ABC, abstractmethod


class Payment(ABC):

    @abstractmethod
    def process_payment(self, amount):
        pass


class UPI(Payment):
    def process_payment(self, amount):
        print("Payment of", amount, "processed using UPI")


class CreditCard(Payment):
    def process_payment(self, amount):
        print("Payment of", amount, "processed using Credit Card")


class NetBanking(Payment):
    def process_payment(self, amount):
        print("Payment of", amount, "processed using Net Banking")


upi = UPI()
credit_card = CreditCard()
net_banking = NetBanking()

upi.process_payment(500)
credit_card.process_payment(1000)
net_banking.process_payment(2000)
```

```python
from abc import ABC, abstractmethod


class FoodOrder(ABC):

    @abstractmethod
    def place_order(self):
        pass

    @abstractmethod
    def calculate_bill(self):
        pass


class PizzaOrder(FoodOrder):
    def place_order(self):
        print("Pizza order placed")

    def calculate_bill(self):
        return 300


class BurgerOrder(FoodOrder):
    def place_order(self):
        print("Burger order placed")

    def calculate_bill(self):
        return 200


class BiryaniOrder(FoodOrder):
    def place_order(self):
        print("Biryani order placed")

    def calculate_bill(self):
        return 250


pizza = PizzaOrder()
burger = BurgerOrder()
biryani = BiryaniOrder()

pizza.place_order()
print("Pizza Bill:", pizza.calculate_bill())

burger.place_order()
print("Burger Bill:", burger.calculate_bill())

biryani.place_order()
print("Biryani Bill:", biryani.calculate_bill())
```

```python
from abc import ABC, abstractmethod


class Ride(ABC):

    def __init__(self, distance):
        self.distance = distance

    @abstractmethod
    def calculate_fare(self):
        pass


class BikeRide(Ride):
    def calculate_fare(self):
        return self.distance * 10


class AutoRide(Ride):
    def calculate_fare(self):
        return self.distance * 15


class CabRide(Ride):
    def calculate_fare(self):
        return self.distance * 20


bike = BikeRide(10)
auto = AutoRide(10)
cab = CabRide(10)

print("Bike Fare:", bike.calculate_fare())
print("Auto Fare:", auto.calculate_fare())
print("Cab Fare:", cab.calculate_fare())
```

```python
from abc import ABC, abstractmethod


class Product(ABC):

    def __init__(self, product_name, product_id, price):
        self.product_name = product_name
        self.product_id = product_id
        self.__price = price

    def set_price(self, price):
        if price > 0:
            self.__price = price
            print("Price updated successfully")
        else:
            print("Price must be positive")

    def get_price(self):
        return self.__price

    @abstractmethod
    def calculate_discount(self):
        pass

    @abstractmethod
    def calculate_final_price(self):
        pass


class Electronics(Product):

    def calculate_discount(self):
        return self.get_price() * 10 / 100

    def calculate_final_price(self):
        discount = self.calculate_discount()
        return self.get_price() - discount


class Clothing(Product):

    def calculate_discount(self):
        return self.get_price() * 20 / 100

    def calculate_final_price(self):
        discount = self.calculate_discount()
        return self.get_price() - discount


class Grocery(Product):

    def calculate_discount(self):
        return self.get_price() * 5 / 100

    def calculate_final_price(self):
        discount = self.calculate_discount()
        return self.get_price() - discount


product1 = Electronics("Laptop", 101, 50000)
product2 = Clothing("Shirt", 102, 2000)
product3 = Grocery("Rice", 103, 1000)

print("Laptop Final Price:", product1.calculate_final_price())
print("Shirt Final Price:", product2.calculate_final_price())
print("Rice Final Price:", product3.calculate_final_price())
```

```python
from abc import ABC, abstractmethod


class Person:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def display_role(self):
        print("I am a person")


class Doctor(Person):

    def display_role(self):
        print("I am a Doctor")


class Patient(Person):

    def __init__(self, name, age, medical_record):
        super().__init__(name, age)
        self.__medical_record = medical_record

    def display_role(self):
        print("I am a Patient")

    def get_medical_record(self):
        return self.__medical_record


class Nurse(Person):

    def display_role(self):
        print("I am a Nurse")


class HospitalService(ABC):

    @abstractmethod
    def provide_service(self):
        pass


class Consultation(HospitalService):

    def provide_service(self):
        print("Doctor consultation provided")


class LabTest(HospitalService):

    def provide_service(self):
        print("Lab test provided")


class Pharmacy(HospitalService):

    def provide_service(self):
        print("Medicine provided by pharmacy")


doctor = Doctor("Arun", 40)
patient = Patient("Dharun", 24, "Blood Test Report")
nurse = Nurse("Priya", 30)

doctor.display_role()
patient.display_role()
nurse.display_role()

print("Medical Record:", patient.get_medical_record())

consultation = Consultation()
lab_test = LabTest()
pharmacy = Pharmacy()

consultation.provide_service()
lab_test.provide_service()
pharmacy.provide_service()
```

```python
from abc import ABC, abstractmethod


class BankAccount(ABC):

    def __init__(self, name, account_number, balance):
        self.name = name
        self.account_number = account_number
        self.__balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.__balance = self.__balance + amount
            print("Amount deposited successfully")
        else:
            print("Deposit amount must be positive")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.__balance:
            self.__balance = self.__balance - amount
            print("Amount withdrawn successfully")
        else:
            print("Invalid withdrawal")

    def get_balance(self):
        return self.__balance

    @abstractmethod
    def calculate_interest(self):
        pass

    def display_details(self):
        print("Name:", self.name)
        print("Account Number:", self.account_number)
        print("Balance:", self.__balance)


class SavingsAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 5 / 100


class CurrentAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 2 / 100


class SalaryAccount(BankAccount):

    def calculate_interest(self):
        return self.get_balance() * 4 / 100


savings = SavingsAccount("Dharun", 101, 50000)
current = CurrentAccount("Arun", 102, 50000)
salary = SalaryAccount("Kumar", 103, 50000)

print("Savings Account")
savings.deposit(5000)
savings.withdraw(2000)
print("Balance:", savings.get_balance())
print("Interest:", savings.calculate_interest())
savings.display_details()

print("\nCurrent Account")
current.deposit(5000)
print("Balance:", current.get_balance())
print("Interest:", current.calculate_interest())
current.display_details()

print("\nSalary Account")
salary.deposit(5000)
print("Balance:", salary.get_balance())
print("Interest:", salary.calculate_interest())
salary.display_details()
```

```python
from abc import ABC, abstractmethod


class Employee(ABC):

    def __init__(self, name, employee_id, salary):
        self.name = name
        self.employee_id = employee_id
        self.__salary = salary

    def get_salary(self):
        return self.__salary

    def set_salary(self, salary):
        if salary >= 0:
            self.__salary = salary
        else:
            print("Salary cannot be negative")

    @abstractmethod
    def calculate_salary(self):
        pass

    @abstractmethod
    def calculate_bonus(self):
        pass

    def salary_slip(self):
        salary = self.calculate_salary()
        bonus = self.calculate_bonus()
        total = salary + bonus

        print("\n----- SALARY SLIP -----")
        print("Name:", self.name)
        print("Employee ID:", self.employee_id)
        print("Salary:", salary)
        print("Bonus:", bonus)
        print("Total Salary:", total)


class FullTimeEmployee(Employee):

    def calculate_salary(self):
        return self.get_salary()

    def calculate_bonus(self):
        return self.get_salary() * 10 / 100


class PartTimeEmployee(Employee):

    def __init__(self, name, employee_id, salary, hours):
        super().__init__(name, employee_id, salary)
        self.hours = hours

    def calculate_salary(self):
        return self.get_salary() * self.hours

    def calculate_bonus(self):
        return self.get_salary() * 5 / 100


class ContractEmployee(Employee):

    def calculate_salary(self):
        return self.get_salary()

    def calculate_bonus(self):
        return self.get_salary() * 3 / 100


full_time = FullTimeEmployee("Dharun", 101, 50000)

part_time = PartTimeEmployee("Arun", 102, 500, 20)

contract = ContractEmployee("Kumar", 103, 40000)


full_time.salary_slip()
part_time.salary_slip()
contract.salary_slip()
```
