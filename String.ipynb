```python
password = input("Enter your password: ")
has_upper = False
has_lower = False
has_digit = False
has_special = False

for char in password:
    if char.isupper():
        has_upper = True
    elif char.islower():
        has_lower = True
    elif char.isdigit():
        has_digit = True
    else:
        has_special = True


if len(password) >= 8 and has_upper and has_lower and has_digit and has_special:
    print("Strong Password")
else:
    print("Weak Password")
```

```python
text = input("Enter a string: ")

duplicates = []

for char in text:
    if text.count(char) > 1 and char not in duplicates:
        duplicates.append(char)

print("Duplicate characters:", duplicates)
```

```python
text = input("Enter a sentence: ")

characters = []

for char in text:
    if char not in characters:
        characters.append(char)

for char in characters:
    print(char, "→", text.count(char))
```

```python
text = input("Enter a word or sentence: ")


text = text.replace(" ", "")
text = text.lower()

reverse = text[::-1]
if text == reverse:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
email = "akshaya@gmail.com"
phone = "9876543210"


email_parts = email.split("@")

username = email_parts[0]
domain = email_parts[1]

masked_email = username[0] + "*" * (len(username) - 1) + "@" + domain

masked_phone = "*" * (len(phone) - 4) + phone[-4:]

print("Masked email:", masked_email)
print("Masked phone:", masked_phone)
```
