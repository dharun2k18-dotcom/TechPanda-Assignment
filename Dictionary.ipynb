```python
sales = {"Laptop": [10, 12, 8],"Mouse": [25, 30, 20],"Keyboard": [15, 18, 12]}
```

```python
for product, monthly_sales in sales.items():
    total = 0
    for sale in monthly_sales:
        total += sale
    print(product, "Total Sales:", total)
```

```python

highest_product = ""
highest_total = 0
for product, monthly_sales in sales.items():
    total = 0
    for sale in monthly_sales:
        total += sale
    if total > highest_total:
        highest_total = total
        highest_product = product
print("Highest sales product:", highest_product)
print("Total sales:", highest_total)
```

```python
total_sales = 0
total_months = 0
for monthly_sales in sales.values():
    for sale in monthly_sales:
        total_sales += sale
        total_months += 1
average = total_sales / total_months

print("Average monthly sales:", average)
```

```python
movies = {"Leo": 4.2,"Vikram": 4.5,"Master": 3.8,"Jailer": 3.8}
```

```python
movies["Kaithi"] = 4.6



movies["Master"] = 4.8




movies.pop("Leo")



highest_movie = ""
highest_rating = 0
for movie, rating in movies.items():
    if rating > highest_rating:
        highest_rating = rating
        highest_movie = movie
print("Highest-rated movie:", highest_movie)
print("Rating:", highest_rating)




lowest_movie = ""
lowest_rating = 5
for movie, rating in movies.items():
    if rating < lowest_rating:
        lowest_rating = rating
        lowest_movie = movie
print("Lowest-rated movie:", lowest_movie)
print("Rating:", lowest_rating)





total = 0
count = 0

for rating in movies.values():
    total += rating
    count += 1
average = total / count
print("Average Rating:", average)




for movie, rating in movies.items():
    if rating > 4:
        print(movie, ":", rating)




movie = "Vikram"
if movie in movies:
    print("Rating:", movies[movie])
else:
    print("Movie not found")
```
