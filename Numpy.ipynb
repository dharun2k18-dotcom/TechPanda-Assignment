```python
salaries = np.array([25000, 32000, 45000, 28000, 60000, 38000])

average = np.mean(salaries)

above_average = salaries[salaries > average]

print("Average salary:", average)
print("Salaries above average:", above_average)
```

```python
sales = np.array([1200, 1500, 900, 2200, 1800, 2500, 1700])

total_sales = np.sum(sales)
average_sales = np.mean(sales)
highest_sales = np.max(sales)
highest_day = np.argmax(sales) + 1

print("Total sales:", total_sales)
print("Average sales:", average_sales)
print("Highest sales:", highest_sales)
print("Day with highest sales:", highest_day)
```

```python
marks = np.array([78, 92, 35, 84, 67, 29, 95, 40])

high_scores = marks[marks >= 80]
passed_count = np.sum(marks >= 40)

print("Marks of students scoring at least 80:", high_scores)
print("Number of students who passed:", passed_count)
```

```python
arr = np.array([10, 20, 30, 40, 50])

normalized = (arr - np.min(arr)) / (np.max(arr) - np.min(arr))

print(normalized)
```

```python
sales = np.array([[100, 150, 200],
                  [120, 180, 160],
                  [190, 140, 210],
                  [200, 170, 190]])

product_sales = np.sum(sales, axis=1)
monthly_sales = np.sum(sales, axis=0)

highest_month = np.argmax(monthly_sales) + 1
highest_month_sales = np.max(monthly_sales)

print("Total sales for each product:", product_sales)
print("Total sales for each month:", monthly_sales)
print("Month with highest sales:", highest_month)
print("Highest monthly sales:", highest_month_sales)
```

```python
arr = np.random.randint(1, 101, 20)

mean_value = np.mean(arr)
median_value = np.median(arr)
unique_values = np.unique(arr)
even_numbers = arr[arr % 2 == 0]
largest_three = np.sort(arr)[-3:][::-1]

print("Random array:", arr)
print("Mean:", mean_value)
print("Median:", median_value)
print("Unique values:", unique_values)
print("Even numbers:", even_numbers)
print("Three largest values:", largest_three)
```
