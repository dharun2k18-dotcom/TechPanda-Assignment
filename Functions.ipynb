```python
def total_transactions(transactions):
    return sum(transactions)


def deposit(transactions):
    deposit = 0

    for amount in transactions:
        if amount > 0:
            deposit = deposit + amount

    return deposit


def withdraw(transactions):
    withdraw = 0

    for amount in transactions:
        if amount < 0:
            withdraw = withdraw + amount

    return withdraw


def transaction_count(transactions):
    return len(transactions)


transactions = [500, -200, 1000, -300, 700]
```

```python
def validate_employee(employee):
    if employee["name"] == "":
        return False

    if employee["salary"] <= 0:
        return False

    return True


def calculate_bonus(employee):
    bonus = employee["salary"] * 10 / 100
    return bonus


def format_employee(employee, bonus):
    return employee["name"] + " - Salary: " + str(employee["salary"]) + " - Bonus: " + str(bonus)


def process_employee(employee):

    if validate_employee(employee):

        bonus = calculate_bonus(employee)

        result = format_employee(employee, bonus)

        return result

    else:
        return "Invalid employee"


employee = {"name": "Dharun","salary": 50000}

print(process_employee(employee))
```

```python
def calculate_total(marks):
    return sum(marks)


def calculate_percentage(marks):
    total = calculate_total(marks)
    percentage = total / len(marks)
    return percentage


def calculate_grade(percentage):

    if percentage >= 90:
        return "A"

    elif percentage >= 80:
        return "B"

    elif percentage >= 70:
        return "C"

    elif percentage >= 60:
        return "D"

    else:
        return "E"


def result_status(marks):

    for mark in marks:
        if mark < 40:
            return "Fail"

    return "Pass"


def generate_report(name, marks):

    total = calculate_total(marks)

    percentage = calculate_percentage(marks)

    grade = calculate_grade(percentage)

    status = result_status(marks)

    print("_____STUDENT REPORT_____")
    print("Name:", name)
    print("Marks:", marks)
    print("Total:", total)
    print("Percentage:", percentage)
    print("Grade:", grade)
    print("Result:", status)


name = input("Enter student name: ")

marks = []

for i in range(5):
    mark = int(input("Enter mark: "))
    marks.append(mark)


generate_report(name, marks)
```

```python
balance = 0
transactions = []


def deposit(amount):
    global balance

    balance = balance + amount
    transactions.append("Deposited: " + str(amount))

    print("Amount deposited successfully")


def withdraw(amount):
    global balance

    if amount <= balance:
        balance = balance - amount
        transactions.append("Withdrawn: " + str(amount))
        print("Amount withdrawn successfully")
    else:
        print("Insufficient balance")


def check_balance():
    print("Balance:", balance)


def transaction_history():
    print("Transaction History:")

    for transaction in transactions:
        print(transaction)
```

```python
salary = 50000

def basic_salary(salary):
    return salary


def allowance(salary):
    return salary * 20 / 100


def deduction(salary):
    return salary * 10 / 100


def net_salary(salary):
    basic = basic_salary(salary)
    allowances = allowance(salary)
    deductions = deduction(salary)

    return basic + allowances - deductions


```

```python
print("Basic Salary:", basic_salary(salary))
print("Allowance:", allowance(salary))
print("Deduction:", deduction(salary))
print("Net Salary:", net_salary(salary))
```

```python
cart = []


def add_product(name, price):
    product = {"name": name,"price": price}

    cart.append(product)
    print("Product added")


def remove_product(name):
    for product in cart:
        if product["name"] == name:
            cart.remove(product)
            print("Product removed")
            return

    print("Product not found")


def calculate_subtotal():
    total = 0

    for product in cart:
        total = total + product["price"]

    return total


def apply_discount(total):
    discount = total * 10 / 100
    return total - discount


def generate_bill():
    print("______BILL______")

    for product in cart:
        print(product["name"], "-", product["price"])

    subtotal = calculate_subtotal()
    final_price = apply_discount(subtotal)

    print("Subtotal:", subtotal)
    print("Final Price:", final_price)
```

```python
books = {}


def add_book(book):
    books[book] = "Available"
    print("Book added")


def issue_book(book):
    if book in books:

        if books[book] == "Available":
            books[book] = "Issued"
            print("Book issued")
        else:
            print("Book already issued")

    else:
        print("Book not found")


def return_book(book):
    if book in books:
        books[book] = "Available"
        print("Book returned")
    else:
        print("Book not found")


def check_availability(book):
    if book in books:
        print("Status:", books[book])
    else:
        print("Book not found")
```

```python
expenses = []


def add_expense(name, amount, category):

    expense = {"name": name,"amount": amount,"category": category}

    expenses.append(expense)


def category_total(category):

    total = 0

    for expense in expenses:
        if expense["category"] == category:
            total = total + expense["amount"]

    return total


def calculate_total():

    total = 0

    for expense in expenses:
        total = total + expense["amount"]

    return total


def generate_summary():

    print("----- EXPENSE SUMMARY -----")

    for expense in expenses:
        print(expense["name"],"-",expense["amount"],"-",expense["category"])

    print("Total Expense:", calculate_total())
```

```python
add_expense("Food", 500, "Food")
add_expense("Bus", 100, "Travel")
add_expense("Movie", 300, "Entertainment")
add_expense("Lunch", 250, "Food")
```

```python
generate_summary()
```

```python
print("Food expenses:", category_total("Food"))
```

```python
inventory = {}


def add_stock(product, quantity, price):

    inventory[product] = {"quantity": quantity,"price": price}


def update_stock(product, quantity):

    if product in inventory:
        inventory[product]["quantity"] = quantity
        print("Stock updated")
    else:
        print("Product not found")


def search_product(product):

    if product in inventory:
        print("Product found")
        print("Quantity:", inventory[product]["quantity"])
        print("Price:", inventory[product]["price"])
    else:
        print("Product not found")


def inventory_value():

    total = 0

    for product in inventory:
        quantity = inventory[product]["quantity"]
        price = inventory[product]["price"]

        total = total + (quantity * price)

    return total
```

```python
add_stock("Laptop", 5, 65000)
add_stock("Mouse", 10, 1200)
```

```python
update_stock("Laptop", 7)
```

```python
search_product("Laptop")
```

```python
print("Total Inventory Value:", inventory_value())
```

```python
contacts = {}


def add_contact(name, phone):

    contacts[name] = phone
    print("Contact added")


def search_contact(name):

    if name in contacts:
        print("Phone:", contacts[name])
    else:
        print("Contact not found")


def update_contact(name, phone):

    if name in contacts:
        contacts[name] = phone
        print("Contact updated")
    else:
        print("Contact not found")


def delete_contact(name):

    if name in contacts:
        del contacts[name]
        print("Contact deleted")
    else:
        print("Contact not found")
```

```python
add_contact("Dharun", "9876543210")
add_contact("Arun", "9876543211")
```

```python
search_contact("Dharun")
```

```python
update_contact("Dharun", "9999999999")
```

```python
delete_contact("Arun")
```

```python
search_contact("Dharun")
```

```python
balance = 10000
pin = "1234"
transactions = []


def validate_pin():

    entered_pin = input("Enter PIN: ")

    if entered_pin == pin:
        return True
    else:
        return False


def check_balance():

    print("Balance:", balance)


def withdraw(amount):

    global balance

    if amount <= balance:
        balance = balance - amount

        transactions.append("Withdrawn: "+str(amount))

        print("Withdrawal successful",balance)

    else:
        print("Insufficient balance")


def deposit(amount):

    global balance

    balance = balance + amount

    transactions.append("Deposited: " +str(amount))

    print("Deposit successful",amount)


def show_transactions():

    for transaction in transactions:
        print(transaction)
```

```python
check_balance()
```

```python
deposit(5000)
```

```python
withdraw(2000)
```

```python
check_balance()
```

```python
show_transactions()
```
