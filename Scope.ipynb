```python
cart_total = 0

def add_amount(amount):
    global cart_total
    cart_total = cart_total + amount

def remove_amount(amount):
    global cart_total
    cart_total = cart_total - amount

add_amount(500)
add_amount(300)

print("After adding:", cart_total)

remove_amount(200)

print("After removing:", cart_total)
```

```python
transactions = 0

def deposit():
    global transactions

    amount = int(input("Enter deposit amount: "))

    transactions = transactions + 1

    print("Deposited:", amount)


def withdraw():
    global transactions

    amount = int(input("Enter withdrawal amount: "))

    transactions = transactions + 1

    print("Withdrawn:", amount)


while True:

    print("\n1. Deposit")
    print("2. Withdraw")
    print("3. Show Transactions")
    print("4. Exit")

    choice = int(input("Enter choice(1 for deposit, 2 for withdrawal, 3 for transactions, 4 to exit): "))

    if choice == 1:
        deposit()

    elif choice == 2:
        withdraw()

    elif choice == 3:
        print("Total transactions:", transactions)

    elif choice == 4:
        print("Exiting...")
        break

    else:
        print("Invalid choice")
```

```python
college_name = "SXCCE College"


def student():

    student_name = "Dharun"

    def marks_details():

        marks = 85

        def calculate_grade():

            if marks >= 90:
                grade = "A+"

            elif marks >= 80:
                grade = "A"

            elif marks >= 70:
                grade = "B"

            else:
                grade = "C"

            print("College:", college_name)
            print("Student:", student_name)
            print("Marks:", marks)
            print("Grade:", grade)

        calculate_grade()

    marks_details()


student()
```
