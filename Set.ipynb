```python
python_club = {"Arun", "Priya", "Kavin", "Meena", "Rahul"}
data_club = {"Priya", "Meena", "Sanjay", "Divya"}
```

```python
print("Students in either club:", python_club.union(data_club))


print("Students in both clubs:", python_club.intersection(data_club))


print("Students only in Python Club:", python_club.difference(data_club))


print("Students only in Data Club:", data_club.difference(python_club))


print("Students in exactly one club:", python_club.symmetric_difference(data_club))
```

```python
dance = {"Arun", "Priya", "Kavin", "Meena"}
music = {"Priya", "Rahul", "Meena", "Divya"}
sports = {"Kavin", "Rahul", "Sanjay"}
```

```python
print("Dance and Music:", dance.intersection(music))


print("Only Dance:", dance.difference(music))


print("Music or Sports:", music.union(sports))


print("Exactly one of Dance and Music:",dance.symmetric_difference(music))


print("Dance and Sports are disjoint:",dance.isdisjoint(sports))


all_participants = dance.union(music).union(sports)
print("All participants:", all_participants)


print("Dance is subset of all participants:",dance.issubset(all_participants))


print("All participants are superset of Music:",all_participants.issuperset(music))
```

```python
employee1 = {"Python", "SQL", "Excel"}
employee2 = {"Python", "SQL", "Power BI", "Excel"}
employee3 = {"Java", "C++", "SQL"}
```

```python
python_skills = {"Arun", "Priya", "Kavin", "Meena"}
sql_skills = {"Priya", "Kavin", "Rahul", "Meena"}
data_analysis_skills = {"Kavin", "Meena", "Sanjay"}
```

```python
print("Python and SQL:",python_skills.intersection(sql_skills))


all_three = python_skills.intersection(sql_skills).intersection(data_analysis_skills)
print("All three skills:", all_three)


only_python = python_skills.difference(sql_skills).difference(data_analysis_skills)
print("Only Python:", only_python)


print("Python or Data Analysis:",python_skills.union(data_analysis_skills))


exactly_one = (python_skills.symmetric_difference(sql_skills).symmetric_difference(data_analysis_skills))

print("Exactly one skill:", exactly_one)


print("Python and SQL are disjoint:",python_skills.isdisjoint(sql_skills))


print("All three is subset of Python:",all_three.issubset(python_skills))


print("Python is superset of all three:",python_skills.issuperset(all_three))
```
